# Speech Features Exploration
Notebook pour explorer les features acoustiques et valider les distributions avant entraînement.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

data_dir = Path('..') / 'data'
csv_files = sorted(data_dir.glob('*.csv'))
df = pd.concat([pd.read_csv(p) for p in csv_files], ignore_index=True) if csv_files else pd.DataFrame()
df.head()
```

```python
if not df.empty:
    numeric_cols = ['speech_rate', 'pitch_variation', 'filler_rate', 'energy_level', 'pause_count', 'score']
    df[numeric_cols].hist(figsize=(12, 8), bins=30)
    plt.tight_layout()
    plt.show()
else:
    print('Aucune donnée disponible dans training/data')
```
